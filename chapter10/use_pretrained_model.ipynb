```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer("cosineloss_embedding_model")

sentence1 = "A man is playing guitar"
sentence2 = "A person is playing a musical instrument"
sentence3 = "A cat is sleeping on the sofa"

emb1 = model.encode(sentence1, convert_to_tensor=True)
emb2 = model.encode(sentence2, convert_to_tensor=True)
emb3 = model.encode(sentence3, convert_to_tensor=True)

print(util.cos_sim(emb1, emb2))  # 高相似度（蕴含关系）
print(util.cos_sim(emb1, emb3))  # 低相似度（无关）
```
