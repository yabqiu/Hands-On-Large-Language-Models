```python
import kick_zscaler

from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline, GenerationConfig

# model_name = "microsoft/Phi-3-mini-4k-instruct"
model_name = "google/gemma-4-E4B-it"
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="mps",
)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python

pipe = pipeline(
    "text-generation",
    model=model,
    device_map="mps",
    tokenizer=tokenizer,
    return_full_text=False,
    max_new_tokens=10,
    max_length=None,
    do_sample=False,
)

messages = [
    {"role": "user", "content": "The sky is"}
]

output = pipe(messages)
print(output[0]["generated_text"])
```

```python
prompt = pipe.tokenizer.apply_chat_template(messages, tokenizer=False)
print(pipe.tokenizer.decode(prompt["input_ids"]))
```

```python
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model="google/gemma-4-E4B-it",
    device_map="mps",
    # tokenizer=tokenizer,
    return_full_text=False,
    max_new_tokens=200,
    max_length=None,
    do_sample=False,
)

messages = [
    {"role": "user", "content": "树上有 9 只鸟，猎人开枪打死了一只，树上还剩几只鸟？"},
    {"role": "assistant", "content": "树上还剩 0 只鸟，因为枪声吓跑了其他的鸟。"},
    {"role": "user", "content": "树上有 18 只鸟，开枪打死了两只，树上还剩几只鸟？只用一个数字回答, 并解释原因"}
]

output = pipe(messages)
print(output[0]["generated_text"])
```
